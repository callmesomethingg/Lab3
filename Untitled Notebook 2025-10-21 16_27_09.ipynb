```python
from pyspark.sql.functions import lit, monotonically_increasing_id

# ===============================================================
# Step 1️⃣: Load Silver-layer datasets
# ===============================================================
books = spark.read.parquet("abfss://lakehouse@goodreadreviews60312547.dfs.core.windows.net/processed/books/")
authors = spark.read.parquet("abfss://lakehouse@goodreadreviews60312547.dfs.core.windows.net/processed/authors/")
reviews_clean = spark.read.parquet("abfss://lakehouse@goodreadreviews60312547.dfs.core.windows.net/processed/reviews/")

# ===============================================================
# Step 2️⃣: Create simulated book_authors bridge
# ===============================================================
book_authors = books.select("book_id").withColumn("author_id", monotonically_increasing_id())

# ===============================================================
# Step 3️⃣: Rename overlapping columns to avoid ambiguity
# ===============================================================
books = (
    books.withColumnRenamed("average_rating", "book_avg_rating")
         .withColumnRenamed("ratings_count", "book_ratings_count")
)

authors = (
    authors.withColumnRenamed("average_rating", "author_avg_rating")
           .withColumnRenamed("ratings_count", "author_ratings_count")
)

# ===============================================================
# Step 4️⃣: Join all DataFrames
# ===============================================================
curated_reviews = (
    reviews_clean
    .join(books, on="book_id", how="inner")
    .join(book_authors, on="book_id", how="inner")
    .join(authors, on="author_id", how="left")
)

# ===============================================================
# Step 5️⃣: Handle language field safely
# ===============================================================
if "language_code" in curated_reviews.columns:
    curated_reviews = curated_reviews.withColumnRenamed("language_code", "language")
else:
    curated_reviews = curated_reviews.withColumn("language", lit("Unknown"))

# ===============================================================
# Step 6️⃣: Select only required columns (no ambiguity now!)
# ===============================================================
curated_reviews = curated_reviews.select(
    "review_id",
    "book_id",
    "title",
    "author_id",
    "name",
    "user_id",
    "rating",
    "review_text",
    "language",
    "book_avg_rating",
    "book_ratings_count"
)

# ===============================================================
# Step 7️⃣: Verify output and save
# ===============================================================
curated_reviews.printSchema()
curated_reviews.show(5, truncate=False)

# Write curated DataFrame to Gold zone as Delta table
curated_reviews.write.format("delta").mode("overwrite").save(
    "abfss://lakehouse@goodreadreviews60312547.dfs.core.windows.net/gold/curated_reviews/"
)

# Register as SQL-accessible table
curated_reviews.write.format("delta").mode("overwrite").saveAsTable("curated_reviews")

# ===============================================================
# Step 8️⃣: Verify in SQL
# ===============================================================
spark.sql("SHOW TABLES").show()
spark.sql("SELECT * FROM curated_reviews LIMIT 5").show(truncate=False)
```

```python
%sql
SELECT 
    title,
    ROUND(AVG(book_avg_rating), 2) AS avg_book_rating,
    COUNT(review_id) AS total_reviews
FROM curated_reviews
GROUP BY title
ORDER BY avg_book_rating DESC
LIMIT 5;
```

```python
%sql
SELECT 
    name AS author_name,
    COUNT(review_id) AS total_reviews,
    ROUND(AVG(book_avg_rating), 2) AS avg_rating
FROM curated_reviews
GROUP BY name
ORDER BY total_reviews DESC
LIMIT 5;
```
