```python
import re
import emoji
from pyspark.sql.functions import udf, col
from pyspark.sql.types import StringType

def clean_text(text):
    if text is None:
        return ""
    text = text.lower()
    text = emoji.replace_emoji(text, "")
    text = re.sub(r"http\S+", " ", text)          # remove URLs
    text = re.sub(r"[^a-z\s]", " ", text)         # keep only letters
    text = re.sub(r"\s+", " ", text).strip()      # remove extra spaces
    return text

clean_udf = udf(clean_text, StringType())
df_clean = df.withColumn("cleaned_review", clean_udf(col("review_text")))

display(df_clean.select("review_text", "cleaned_review").limit(5))
```

```python
from pyspark.sql.functions import length, size, split

df_clean = df_clean.withColumn("review_length_chars", length(col("cleaned_review")))
df_clean = df_clean.withColumn("review_length_words", size(split(col("cleaned_review"), " ")))

display(df_clean.select("cleaned_review", "review_length_chars", "review_length_words").limit(5))
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
from pyspark.sql.functions import pandas_udf, PandasUDFType
from pyspark.sql.types import FloatType
import pandas as pd

sia = SentimentIntensityAnalyzer()

@pandas_udf("float")
def sentiment_compound_udf(texts: pd.Series) -> pd.Series:
    return texts.fillna("").apply(lambda x: sia.polarity_scores(x)["compound"])

df_sent = df_clean.withColumn("sentiment_score", sentiment_compound_udf(col("cleaned_review")))

display(df_sent.select("cleaned_review", "sentiment_score").limit(5))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Collect sample (or convert entire column if small enough)
texts = [row.cleaned_review for row in df_sent.select("cleaned_review").limit(5000).collect()]

vectorizer = TfidfVectorizer(max_features=1000, stop_words="english", ngram_range=(1,2))
tfidf_matrix = vectorizer.fit_transform(texts)
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())

print(tfidf_df.shape)
tfidf_df.head()
```
